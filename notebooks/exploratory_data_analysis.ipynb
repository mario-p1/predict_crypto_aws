# Build simple features

```python
features = (
    df[["timestamp", "close"]]
    .copy()
    .rename(columns={"close": "y"})
    .sort_values("timestamp")
    .reset_index(drop=True)
)
```

```python
features["lag_1"] = features["y"].shift(1)
features["lag_5"] = features["y"].shift(5)
features["lag_10"] = features["y"].shift(10)

features = features.dropna().reset_index(drop=True)
```

```python
features.head(15)
```

```python
from datetime import UTC, datetime

import xgboost as xgb
```

```python
valid_date = datetime(2025, 6, 1, tzinfo=UTC).timestamp() * 1000
test_date = datetime(2026, 1, 1, tzinfo=UTC).timestamp() * 1000


train_df = features[features["timestamp"] < valid_date]
valid_df = features[
    (features["timestamp"] >= valid_date) & (features["timestamp"] < test_date)
]
test_df = features[features["timestamp"] >= test_date]

dtrain = xgb.DMatrix(train_df[["lag_1", "lag_5", "lag_10"]], label=train_df["y"])
dvalid = xgb.DMatrix(valid_df[["lag_1", "lag_5", "lag_10"]], label=valid_df["y"])
dtest = xgb.DMatrix(test_df[["lag_1", "lag_5", "lag_10"]], label=test_df["y"])
```

```python
model = xgb.train(
    dtrain=dtrain,
    params={},
    evals=[(dtrain, "train"), (dvalid, "valid")],
)
```
